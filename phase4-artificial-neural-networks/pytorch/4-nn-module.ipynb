# Pytorch Training Pipeline Improvement Using NN Module and torch.optim

```python
import numpy as np
import pandas as pd

import torch
import torch.nn as nn

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

## 1. Data loading

- Breast Cancer Detection Dataset

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace= True)
```

```python
df.info()
```

## 2. Data Preprocessing

- Train-test split
- Scaling
- Label encoding
- Tensor conversion

### a. Train-test split

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

### b. Scaling

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

### c. Label encoding

```python
y_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

### d. Numpy array to PyTorch tensors

```python
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
X_train_tensor.shape
```

```python
y_train_tensor.shape
```

## 3. Defining The Model

```python
class MySimpleNN(nn.Module):

    def __init__(self, num_features):

        super().__init__()
        self.linear = nn.Linear(num_features, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, X):
        out = self.linear(X)
        return self.sigmoid(out)
```

## 4. Training Pipeline

```python
from torchinfo import summary
```

```python
# create model
model = MySimpleNN(X_train_tensor.shape[1])
summary(model, input_size=(455, 30))
```

```python
# Important Parameters

learning_rate = 0.1
epochs = 25
```

```python
# define loss function
loss_function = nn.BCELoss()
```

```python
# define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)
```

```python
# define loop
for epoch in range(epochs):

    # forward pass
    y_pred = model(X_train_tensor)

    # loss calculate
    loss = loss_function(y_pred, y_train_tensor.reshape(-1, 1))

    # backward pass
    loss.backward()

    # clear gradients
    optimizer.zero_grad()

    # parameters update
    optimizer.step()

    # print loss in each epoch
    print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

```python
model.linear.weight
```

```python
model.linear.bias
```

## 5. Evaluation

```python
# model evaluation
with torch.no_grad():
    y_pred = model.forward(X_test_tensor)
    y_pred = (y_pred > 0.9).float()
    accuracy = (y_pred == y_test_tensor).float().mean()
    print(f'Accuracy: {accuracy.item()}')
```
