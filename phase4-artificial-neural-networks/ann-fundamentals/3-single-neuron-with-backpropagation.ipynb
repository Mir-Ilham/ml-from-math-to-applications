# Single Neuron With Backpropagation - Implementation From Scratch

```python
import numpy as np

class Neuron:
    
    def __init__(self, input_size):
        self.weights = np.ones(input_size, dtype=np.float32)
        # self.weights = np.array([0.1, -0.2])
        self.bias = 0.0
        self.error = []

    def numpy_array(self, x):
        return x if isinstance(x, np.ndarray) else np.array(x)
    
    def sigmoid(self, y: np.ndarray) -> np.ndarray:
        y = numpy_array(y)
        return 1 / (1 + np.exp(-y))
    
    def sigmoid_derivative(self, y: np.ndarray) -> np.ndarray:
        return np.exp(-y) / ((1 + np.exp(-y)) ** 2)

    def fit(self, features: list[list[float]], labels: list[int], learning_rate: float, epochs: int):
        features = numpy_array(features)
        labels = numpy_array(labels)
    
        n = len(features)
    
        for epoch in range(epochs):
            y = features @ self.weights + self.bias
            # Activation function
            z = self.sigmoid(y)
    
            # Error function (Mean Squared Error)
            error = z - labels
            self.error.append(np.mean(error ** 2).item())
    
            # Back propagation
    
            # Intermediate gradients
            dz_dy = sigmoid_derivative(y)
    
            # Gradient of error wrt weights
            de_dw  = 2/n * ((error * dz_dy) @ features)

            # Gradient of error wrt bias
            de_db = 2 * np.mean(error * dz_dy)

            # Coefficients updates
            self.weights -= learning_rate * de_dw
            self.bias -= learning_rate * de_db

    def predict(self, features: list[list[float]]) -> np.ndarray:
        features = numpy_array(features)
        y = features @ self.weights + self.bias
        # Activation function
        z = self.sigmoid(y)
        return z
```

```python
n = Neuron(2)
```

```python
n.fit([[1.0, 2.0], [2.0, 1.0], [-1.0, -2.0]], [1, 0, 0], 0.1, 2)
```
