### Note on Regression Metrics

- Mean Absolute Error: 1/n * ∑ |yi - y_predi|
- Mean Squared Error: 1/n * ∑ (yi - y_predi)^2
- Root Mean Squared Error: sqrt(1/n * ∑ (yi - y_predi)^2)
- R2 Score = 1 - (SSR / SSM), where SSR = sum of squared error of regression line and SSM = sum of squared error of mean line
- A higher R2 Score indicates a better fit with lesser error
- Adjusted R2 Score is a modification of R2 Score introduced to handle cases of irrelevant columns.
- Adjusted R2 = 1 - [(1 - R2)(n - 1) / (n - 1 - k)], where n = # of rows, k = # of independent columns

```python
print("MAE", mean_absolute_error(y_test, y_pred))
print("MSE", mean_squared_error(y_test, y_pred))
print("R2 score", r2_score(y_test, y_pred))
```

### From scratch implementation

- Ordinary Least Square Formula: β=((X⊤X)^−1)X⊤y

```python
class MyLR:

    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        X_train = np.insert(X_train,0,1,axis=1)
        betas = np.linalg.inv(X_train.T @ X_train) @ X_train.T @ y_train
        self.intercept_ = betas[0]
        self.coef_ = betas[1:]

    def predict(self, X_test):
        return (X_test @ self.coef_) + self.intercept_
```

```python
lr = MyLR()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
y_pred = lr.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```
