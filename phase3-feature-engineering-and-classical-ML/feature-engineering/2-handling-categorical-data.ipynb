# Handling Categorical Data

- Data can be numerical or categorical.
- Categorical data can be further divided into **nominal** and **ordinal**.
- **Ordinal Encoding** is used for ordinal data.
- **One-hot Encoding** is used for nominal data.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('customer.csv')
```

```python
df.sample(5)
```

```python
df = df.iloc[:,2:]
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,0:2], df.iloc[:,-1], test_size=0.2)
```

## 1. Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder(categories=[['Poor', 'Average', 'Good'], ['School', 'UG', 'PG']])
```

```python
oe.fit(X_train)
```

```python
X_train = oe.transform(X_train)
```

```python
X_test = oe.transform(X_test)
```

```python
X_test[:5]
```

```python
oe.categories_
```

### Label Encoder

- Label encoder is used for ordinal encoding of data labels

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
le.fit(y_train)
```

```python
le.classes_
```

```python
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```

## 2. OneHotEncoding

```python
df = pd.read_csv('cars.csv')
```

```python
df.head()
```

```python
df['brand'].value_counts()
```

```python
df['brand'].nunique()
```

### OHE using Pandas

```python
pd.get_dummies(df, columns=['fuel', 'owner'])
```

```python
# Solving multicollinearity

pd.get_dummies(df, columns=['fuel', 'owner'], drop_first=True)
```

### OHE using Sklearn

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 0:4], df.iloc[:,-1], test_size=0.2, random_state=42)
```

```python
X_train.head()
```

```python
y_train
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder(drop='first')
```

```python
X_train_new = ohe.fit_transform(X_train[['fuel', 'owner']]).toarray()
```

```python
X_train_new
```

```python
X_test_new = ohe.fit_transform(X_test[['fuel', 'owner']]).toarray()
```

```python
X_test_new
```

```python
np.hstack((X_train[['brand', 'km_driven']].values, X_train_new)).shape
```

```python
np.hstack((X_test[['brand', 'km_driven']].values, X_test_new)).shape
```

### OHE with Top Categories

```python
counts = df['brand'].value_counts()
```

```python
threshold = 100
```

```python
repl = counts[counts <= threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl, 'uncommon'))
```
